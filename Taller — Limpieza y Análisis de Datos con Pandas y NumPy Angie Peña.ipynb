```python
for column in df.columns:
    print(f"--- Unique values for column: {column} ---")
    print(df[column].unique())
    print("\n")
```

```python
# Corregir columna 'cantidad': Reemplazar texto 'three' y convertir a numérico
df['cantidad'] = df['cantidad'].replace('three', 3) # Replace the string 'three' with the number 3
df['cantidad'] = pd.to_numeric(df['cantidad'], errors='coerce') # Convert to numeric, coercing errors to NaN

# Corregir columna 'fecha': Convertir a tipo datetime
df['fecha'] = pd.to_datetime(df['fecha'])

# Estandarizar columna 'pais': Convertir a minúsculas y luego capitalizar la primera letra
df['pais'] = df['pais'].str.lower().str.capitalize()

# Estandarizar columna 'metodo_pago': Convertir a minúsculas y luego capitalizar la primera letra
df['metodo_pago'] = df['metodo_pago'].str.lower().str.capitalize()

# Verificar los cambios con df.info() y los valores únicos de las columnas modificadas
print("\n--- Información del DataFrame después de la corrección de tipos y estandarización ---")
df.info()

print("\n--- Valores únicos de 'cantidad' después de la corrección ---")
print(df['cantidad'].unique())

print("\n--- Valores únicos de 'pais' después de la estandarización ---")
print(df['pais'].unique())

print("\n--- Valores únicos de 'metodo_pago' después de la estandarización ---")
print(df['metodo_pago'].unique())

# Mostrar las primeras filas para verificar los cambios
print("\n--- Primeras 5 filas del DataFrame después de la limpieza ---")
display(df.head())
```

```python
# Rellenar valores nulos en 'precio' con la mediana
median_precio = df['precio'].median()
df['precio'].fillna(median_precio, inplace=True)

# Rellenar valores nulos en 'cantidad' con la mediana
# Aseguramos que la cantidad sea un entero si es apropiado para el contexto
median_cantidad = df['cantidad'].median()
df['cantidad'].fillna(median_cantidad, inplace=True)

# Convertir 'cantidad' a entero después de rellenar los nulos, si tiene sentido para el contexto (ej. no se venden 2.5 unidades)
df['cantidad'] = df['cantidad'].astype(int)

# Covertir 'Col' a Colombia
df['pais'] = df['pais'].replace('Col', 'Colombia')

# Verificar la ausencia de valores nulos
print("--- Conteo de datos faltantes después de rellenar ---")
print(df.isnull().sum())

# Mostrar las primeras filas para verificar los cambios
print("\n--- Primeras 5 filas del DataFrame después de rellenar nulos ---")
display(df.head())
```

Parte 4 — Introducción a NumPy <br>
Actividades <br>
• Convertir columnas numéricas a un array de NumPy:  data = df[['precio',
'cantidad']].to_numpy() <br>
• Separar columnas utilizando indexación:  precios = data[:, 0]
cantidades = data[:, 1] <br>
• Aplicar vectorización para calcular ventas totales: totales = precios *
cantidades<br>

```python
from numpy import array
data = df[['precio', 'cantidad']].to_numpy()
array_precios = data[:, 0]
array_cantidades = data[:, 1]
ventas_totales = array_precios * array_cantidades

print("Ventas totales por cada fila:")
print(ventas_totales)
#


```

Parte 5 — Análisis con NumPy
Utilizando funciones de NumPy, calcular:
• Suma total de ventas  
• Promedio de ventas  
• Venta máxima  
• Cantidad de ventas superiores a 1000  
•  
Debes responder:
¿Qué ventajas observas al usar NumPy?  
¿Qué significa “vectorización”?  
¿Qué hace la expresión data[:, 0]?  

```python
total_ventas = ventas_totales.sum()
print(f"Suma total de ventas: ${total_ventas}")
promedio_ventas = ventas_totales.mean()
print(f"Promedio de ventas: ${promedio_ventas}")
venta_maxima = ventas_totales.max()
print(f"Venta máxima: ${venta_maxima}")
cantidad_ventas_superiores_1000 = (ventas_totales > 1000).sum()
print(f"Cantidad de ventas superiores a 1000: {cantidad_ventas_superiores_1000}")
#
```

¿Qué ventajas observas al usar NumPy?  
Saca datos más rápido y pocas líneas de código
¿Qué significa “vectorización”?
se refiere al proceso de convertir operaciones que normalmente se realizan en elementos individuales de datos en operaciones que se pueden aplicar a matrices
¿Qué hace la expresión data[:, 0]?  
Para decir que quiero tomar de inicio a fin la primera columna de la matriz
