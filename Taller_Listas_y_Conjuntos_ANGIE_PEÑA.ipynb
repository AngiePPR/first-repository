Ejercicio 1: Clientes Duplicados y Segmentación

Contexto:

Tienes una lista de clientes con datos repetidos. Además, cuentas con dos listas: clientes
activos e inactivos.

Objetivos:
1. Eliminar duplicados de la lista original.
2. Identificar clientes en ambos estados.
3. Encontrar clientes solo activos.
4. Recorrer los clientes y clasificarlos.

Pistas:

1. Usa set() para eliminar duplicados.
2. Usa intersection() para encontrar coincidencias.
3. Usa difference() para encontrar diferencias.
4. Usa un for con if para clasificar.

```python
# 1. Creamos y llenamos las listas con datos de prueba
lista_original = ["Ana", "Luis", "Ana", "Pedro", "Marta", "Luis", "Elena"]
clientes_activos = ["Ana", "Pedro", "Elena", "Roberto"]
clientes_inactivos = ["Luis", "Marta", "Ana"] # Ana aparece aquí para el ejemplo de duplicados

# --- OBJETIVO 1: Eliminar duplicados de la lista original ---
# Usamos set() para limpiar la lista original
clientes_unicos = set(lista_original)
print(f"Clientes únicos: {clientes_unicos}")

# --- OBJETIVO 2: Identificar clientes en ambos estados (Intersección) ---
# Transformamos las otras listas a sets para poder operar
set_activos = set(clientes_activos)
set_inactivos = set(clientes_inactivos)

clientes_ambos_estados = set_activos.intersection(set_inactivos)
print(f"Clientes en ambos estados: {clientes_ambos_estados}")

# --- OBJETIVO 3: Encontrar clientes SOLO activos (Diferencia) ---
clientes_solo_activos = set_activos.difference(set_inactivos)
print(f"Clientes solo activos: {clientes_solo_activos}")

# --- OBJETIVO 4: Recorrer y clasificar ---
print("\n--- Clasificación Final ---")
for cliente in clientes_unicos:
    if cliente in set_activos and cliente in set_inactivos:
        print(f" {cliente}: Requiere revisión (Duplicado en estados)")
    elif cliente in set_activos:
        print(f" {cliente}: Cliente Activo")
    elif cliente in set_inactivos:
        print(f" {cliente}: Cliente Inactivo")
    else:
        print(f" {cliente}: Sin estado asignado")
```

Ejercicio 2: Productos y Análisis de Inventario

Contexto:
Tienes una lista de productos vendidos y otra de inventario disponible.

Objetivos:

1. Identificar todos los productos.
2. Encontrar productos vendidos que están en inventario.
3. Detectar productos que no se han vendido.
4. Evaluar productos con una condición usando while.


Pistas:

1. Usa union() para combinar conjuntos.
2. Usa & para intersección.
3. Usa - para diferencia.
4. Convierte el set a lista para usar while.



```python
Productos=["camisa","pantalon","medias","zapatos","falda"]
Inventario=["camisa","pantalon","zapatos"]
Ventas=["camisa","medias","falda"]

print(Productos)
print(Inventario)
print(Ventas)

productosVendidos = set(Inventario) & set(Ventas)
print(f"Productos vendidos: {productosVendidos}")

productosnoVendidos = set(Productos) - set(Ventas)
print(f"Productos no vendidos: {productosnoVendidos}")

lista_productos = list(Productos)

i = 0
while i < len(lista_productos):
    producto = lista_productos[i]

    if producto in Ventas:
        print(producto, "se ha vendido")
    else:
        print(producto, "NO se ha vendido")

    i += 1



```

Ejercicio 3: Estudiantes y Cursos.

Contexto:

Tienes estudiantes inscritos en dos cursos distintos.

Objetivos:

1. Encontrar estudiantes en ambos cursos.
2. Identificar estudiantes exclusivos por curso.
3. Unir todos los estudiantes.
4. Clasificar estudiantes según su participación.

Pistas:

1. Usa & para intersección.
2. Usa difference() para exclusivos.
3. Usa union() para todos.
4. Usa for + if para clasificar.

```python
curso1 = {"Ana", "Luis", "Pedro", "Maria"}
curso2 = {"Luis", "Carlos", "Maria", "Sofia"}


ambos = curso1 & curso2
print("En ambos cursos:", ambos)


solo_curso1 = curso1.difference(curso2)
solo_curso2 = curso2.difference(curso1)

print("Solo en curso 1:", solo_curso1)
print("Solo en curso 2:", solo_curso2)


todos = curso1.union(curso2)
print("Todos los estudiantes:", todos)


print("\nClasificación de estudiantes:")

for estudiante in todos:
    if estudiante in curso1 and estudiante in curso2:
        print(estudiante, "está en ambos cursos")
    elif estudiante in curso1:
        print(estudiante, "solo está en curso 1")
    else:
        print(estudiante, "solo está en curso 2")


```
